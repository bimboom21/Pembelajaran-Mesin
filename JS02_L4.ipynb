```python
import cv2
import matplotlib.pyplot as plt

# Masukkan lokasi file gambar
image_path = 'nama_gambar.jpg'

# Baca gambar
img = cv2.imread(image_path)

# Cek apakah gambar berhasil dibaca
if img is None:
    print("Gambar tidak ditemukan. Periksa nama/path file.")
else:
    # Convert BGR → RGB
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Display
    plt.imshow(img_rgb)
    plt.axis('off')
    plt.show()
```

```python
# Read image (BGR default in OpenCV)
img = cv2.imread("Lenna.png")

# Convert BGR → RGB (to ensure correct color display in matplotlib)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Display
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
img_resized = cv2.resize(img, (300, 300))

img_gray = cv2.cvtColor(img_resized, cv2.COLOR_BGR2GRAY)

plt.imshow(img_gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```
